# Data Cleaning with Pandas 

```python
df = pd.read_excel(r"C:\pandas tutorial\Customer Call List.xlsx")
df
```

```python
pd.set_option('display.max.columns', 20)
pd.set_option('display.max.row', 21)
```

```python
df['Last_Name'] = df['Last_Name'].str.strip('/.124_')
df
```

```python
df['Phone_Number'] = df['Phone_Number'].apply(lambda x: str(x))
df
```

```python
df['Phone_Number'] = df['Phone_Number'].str.replace('[^a-zA-Z0-9^]', '', regex= True)
df
```

```python
df['Phone_Number'] = df['Phone_Number'].apply(lambda x: x[0:3] + '-' + x[3:6] + '-' + x[6:10])
df
```

```python
df['Phone_Number'] = df['Phone_Number'].replace('Na--', '')
df['Phone_Number'] = df['Phone_Number'].replace('nan--', '')
df
```

```python
df[['street_address','state', 'zip_code']] = df['Address'].str.split(',', n= 2, expand= True)
df
```

```python
df = df.drop(columns= 'Not_Useful_Column')
df
```

```python
df['Paying Customer'] = df['Paying Customer'].replace('Yes', 'Y')
df['Paying Customer'] = df['Paying Customer'].replace('No', 'N')
df
```

```python
df['Do_Not_Contact'] = df['Do_Not_Contact'].replace('Yes', 'Y')
df['Do_Not_Contact'] = df['Do_Not_Contact'].replace('No', 'N')
df
```

```python
df = df.replace('N/a', '')
df
```

```python
df = df.fillna('')
df
```

```python
for x in df.index:
    if df.loc[x, 'Phone_Number'] == '':
        df.drop(x, inplace= True)
df        
```

```python
df.reset_index(inplace= True)
df
```

```python
df.drop_duplicates()
df
```

```python
df = df.drop(columns= 'index')
df
```
